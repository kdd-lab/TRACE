```python
import json
import os
```

```python
def structure_json(json_file_path):
     with open(json_file_path, 'r') as file:
        json_data=json.load(file)
        new_json={}
        new_json['features']=[]
        #new_json['dd_pred']=json_data['dd_pred']
        #new_json['bb_pred']=json_data['bb_pred']
        new_json['dd_pred']=1
        new_json['bb_pred']=1


    # feature instances
        i=0
        for numerical_cat, eda in json_data['distribution']['numerical'].items():
            temp_feat = {}  # Create a new dictionary for each feature
            temp_feat['name']=numerical_cat
            temp_feat['rname']=numerical_cat
            temp_feat['eda']= {
                'min': eda['min'],
                'max': eda['max'],
                'q1': eda['1st-quantile'],
                'median': eda['2nd-quantile'],
                'q3': eda['3rd-quantile'],
                'mean': eda['mean'],
                'std': eda['std']
            }
            temp_feat['type']='numeric'
            temp_feat['feature_importance'] = json_data['feature_importance_all'][numerical_cat]
            temp_feat['rule']=[]
            temp_feat['crules']={}
            temp_feat['instance_value']=json_data['inst'][numerical_cat]
            new_json['features'].append(temp_feat)

         
        for categorical_cat, categorical_cat_dict in json_data['distribution']['categorical'].items():
            temp_feat_value = json_data['inst'][categorical_cat]
            temp_feat = {}
            
            temp_feat['name']=f"{categorical_cat}={temp_feat_value}"
            temp_feat['rname']=categorical_cat
            temp_feat['eda']= {
                'count': 5,
                'category': 0 
            }
            temp_feat['type']='categorical'
            temp_feat['feature_importance'] = json_data['feature_importance_all'][categorical_cat]
            temp_feat['rule']=[]
            temp_feat['crules']={}
            temp_feat['instance_value']= 1 if temp_feat_value == 0 else 0
            new_json['features'].append(temp_feat)

            temp_feat = {}
            
            temp_feat['name']=f"{categorical_cat}={temp_feat_value}"
            temp_feat['rname']=categorical_cat
            temp_feat['eda']= {
                'count': 5,
                'category': 1 
            }
            temp_feat['type']='categorical'
            temp_feat['feature_importance'] = json_data['feature_importance_all'][categorical_cat]
            temp_feat['rule']=[]
            temp_feat['crules']={}
            temp_feat['instance_value']= 1 if temp_feat_value == 1 else 0
            
            new_json['features'].append(temp_feat)
   
    # rules
        for element in new_json['features']:
            for rule in json_data['rule']:
                if rule['att'] == element['name']:
                    element['rule'] = [[rule,1]]
                    if rule['thr']==0:
                        rule['thr']=0.5
                        rule['op']="<="
                    else:
                        rule['thr']=0.5
                        rule['op']=">="
                        
            #element['rule'] = [[rule, new_json['dd_pred']]]
            #new_json['features'].append(temp_feat)
            
            
         
       # for old_rule_index, rule_dict in json['rule'].items():
            # cat_name = rule_dict['att']
            # pos_cat_name_in_features_list = features.index(cat_name)
            # new_cat_rules = new_json[pos_cat_name_in_features_list]['rule']
            # new_rule={}
            # new_rule['att']= cat_name
            # new_rule['op'] = rule_dict['op']
            # new_rule['thr'] =  rule_dict['thr']
            # new_rule['is_continuous']=rule_dict['is_continuous']
            # new_json[pos_cat_name_in_features_list]['rule'][len(new_cat_rules)][0] = new_rule
            # new_json[pos_cat_name_in_features_list]['rule'][len(new_cat_rules)][1] = 1

    # crules
        
        
        
        
       
    
        return new_json
```
